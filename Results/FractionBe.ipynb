```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_chord_diagram import chord_diagram

# ==========================
# 1. Datos
# ==========================
group_I_params = np.array(
    [
        [3.57, 0.37, 11.70, 1.32, 0.78],  # HD 5394
        [3.54, 0.11, 7.04, 1.32, 0.76],  # HD 20336
        [3.72, -0.12, 8.94, 1.32, 0.73],  # HD 191610
        [4.44, 0.23, 10.99, 1.32, 0.84],  # HD 193009
        [3.76, -0.17, 9.06, 1.32, 0.73],  # HD 194335
        [4.41, 0.24, 11.04, 1.32, 0.80],  # HD 194883
        [3.53, 0.38, 11.87, 1.32, 0.79],  # HD 204722
        [3.60, 0.36, 11.58, 1.32, 0.77],  # HD 202904
        [4.22, 0.37, 11.94, 1.32, 0.82],  # HD 200310
        [3.48, 0.09, 8.64, 1.32, 0.74],  # HD 197419
    ]
)

group_II_params = np.array(
    [
        [3.58, 0.33, 11.79, 1.32, 0.77],  # HD 4180
        [4.40, 0.24, 11.08, 1.32, 0.81],  # HD 11606
        [3.52, 0.37, 11.95, 1.32, 0.79],  # HD 23630
        [3.90, -0.80, 7.21, 1.31, 0.60],  # HD 23862
        [3.80, -0.74, 5.85, 1.31, 0.58],  # HD 217891
        [3.66, -0.06, 6.53, 1.32, 0.71],  # HD 210129
        [4.27, 0.36, 12.02, 1.32, 0.85],  # HD 195907
    ]
)

# Combinar todos los parámetros
params = np.vstack([group_I_params, group_II_params])

# Etiquetas
labels_group_I = [
    'HD 5394',
    'HD 20336',
    'HD 191610',
    'HD 193009',
    'HD 194335',
    'HD 194883',
    'HD 204722',
    'HD 202904',
    'HD 200310',
    'HD 197419',
]

labels_group_II = [
    'HD 4180',
    'HD 11606',
    'HD 23630',
    'HD 23862',
    'HD 217891',
    'HD 210129',
    'HD 195907',
]

labels = labels_group_I + labels_group_II

# ==========================
# 2. Matriz de similitud
# ==========================
n = params.shape[0]
similarity = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i != j:
            d = np.linalg.norm(params[i] - params[j])
            similarity[i, j] = 1.0 / (1.0 + d)  # similitud ∈ (0,1]

# eliminar autoconexiones
np.fill_diagonal(similarity, 0.0)

# Opcional: filtrar conexiones débiles
threshold = 0.25
sim_filtered = similarity.copy()
sim_filtered[sim_filtered < threshold] = 0.0

# ==========================
# 3. Colores por grupo
# ==========================
colors = ['tab:blue'] * len(labels_group_I) + ['tab:orange'] * len(
    labels_group_II
)

# ==========================
# 4. Chord diagram
# ==========================
plt.figure(figsize=(8, 8))
chord_diagram(
    sim_filtered,
    names=labels,
    colors=colors,
    rotate_names=True,
    sort='distance',  # que ordene nodos por conexión
    chordwidth=0.7,
    gap=0.03,
)
#plt.title("Chord diagram – Similaridad de parámetros (Groups I & II)")
plt.show()
```
